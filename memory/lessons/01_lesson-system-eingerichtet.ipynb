## Was ist ein Jupyter-Notebook?

Ein Notebook ist eine Datei (Endung `.ipynb`), die zwei Arten von Zellen mischt:

1. **Markdown-Zellen** — Text wie dieser hier (Erklärungen, Überschriften, Listen)
2. **Code-Zellen** — echter Python-Code, den du **direkt ausführen** kannst

Das ist wie ein Lehrbuch, in dem du jedes Beispiel sofort selbst laufen lassen und ändern kannst.

## Was ist im Projekt Jobetes gerade los?

Der letzte technische Run (Iteration #12) hat **Performance-Budgets** in die CI eingebaut. Im Klartext:

- Wenn jemand Code pusht, der die Web-App **größer als 130 KB** macht, **schlägt der Build fehl**.
- Heute liegt die App bei **104.74 KB** — also 25 KB Puffer.

Warum? Eine größere App lädt langsamer. Bei Patient:innen mit schwachem Mobilfunk in Jordanien zählt jede Kilobyte. Lass uns das Verhältnis ausrechnen:

```python
budget_kb = 130
aktuell_kb = 104.74

puffer = budget_kb - aktuell_kb
auslastung = aktuell_kb / budget_kb * 100

print(f"Puffer: {puffer:.2f} KB")
print(f"Auslastung: {auslastung:.1f}% des Budgets")
```

```python
# Spielwiese: was passiert, wenn die App auf 135 KB waechst?
neue_groesse = 135

if neue_groesse > budget_kb:
    print(f"BUILD FAILS! {neue_groesse} KB > {budget_kb} KB Budget")
else:
    print(f"OK — {neue_groesse} KB unter Budget")
```
